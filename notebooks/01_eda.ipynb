```python
# Summary of the dataset

def dataset_summary(df):
    summary = pd.DataFrame({
        "Data Type": df.dtypes,
        "Missing Values": df.isnull().sum(),
        "Missing Percentage": df.isnull().mean() * 100,
        "Unique Values": df.nunique()
    })
    return summary
```
